```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('hospital_data.db')
df = pd.read_sql_query("SELECT * FROM Full_Analysis_Table", conn)
conn.close()

df.head()
```

```python
df.info()
df.describe()
```

```python
df['Left_Before_Seen_Pct'].describe()
```

```python
df[['ED_Wait_Discharged', 'ED_Overall_Wait', 'Overall_Rating', 'Recommend_Score', 'Nurse_Score', 'Doctor_Score']].corr()
```

```python
corr_matrix = df[['ED_Wait_Discharged', 'ED_Overall_Wait', 'Overall_Rating', 'Recommend_Score', 'Nurse_Score', 'Doctor_Score']].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix, 
    annot = True, #shows coefficients in the box
    fmt='.2f', 
    cmap="coolwarm", 
    vmin=-1, 
    vmax=1, 
    linewidths=0.5, 
    square=True
)

plt.title("Correlation Heatmap", fontsize=16, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```

weak negative correlations between wait time and satisfaction - so as one increase other decrease, and its weak correlation arounf .15-30.

wait time matters, but more factors play a role. 

assuming nurse score has slightly higher correlation due to nurses usually managing patients during the wait. since seeing doctor comes after nurse, ig seen as part of the waiting process before the doctor

```python
# basic scatter plot

plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='ED_Overall_Wait', y='Overall_Rating', alpha=0.4)
plt.xlabel('ED Wait Time (minutes)')
plt.ylabel('Overall Patient Rating')
plt.title('ED Wait Time vs. Overall Patient Satisfaction')
plt.show()
```

```python
sns.regplot(data=df, x='ED_Overall_Wait', y='Overall_Rating', 
            scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.xlabel('ED Wait Time (minutes)')
plt.ylabel('Overall Patient Rating')
plt.title('ED Wait Time vs. Overall Patient Satisfaction')
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18,5))

sns.regplot(data=df, x='ED_Overall_Wait', y='Overall_Rating', ax=axes[0], scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
axes[0].set_title('Wait Time vs Overall Rating')

sns.regplot(data=df, x='ED_Overall_Wait', y='Nurse_Score', ax=axes[1], scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
axes[1].set_title('Wait Time vs Nurse Score')

sns.regplot(data=df, x='ED_Overall_Wait', y='Recommend_Score', ax=axes[2], scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
axes[2].set_title('Wait Time vs Recommend Score')

plt.tight_layout()
plt.show()
```

few that are at the end of ranges, but not exactly any outliers. maybe a couple that are small  outliers. 
the graphs arent showing anything we didnt get from the matrix. overall a relationship, just not super strong. im starting to get curious to work with the rest of the variables and how they would relate to the rest.
