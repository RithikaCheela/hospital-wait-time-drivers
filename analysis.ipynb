```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('hospital_data.db')
df = pd.read_sql_query("SELECT * FROM Full_Analysis_Table", conn)
conn.close()

df.head()
```

```python
df.info()
df.describe()
```

```python
df['Left_Before_Seen_Pct'].describe()
```

```python
df[['ED_Wait_Discharged', 'ED_Overall_Wait', 'Overall_Rating', 'Recommend_Score', 'Nurse_Score', 'Doctor_Score']].corr()
```

```python
corr_matrix = df[['ED_Wait_Discharged', 'ED_Overall_Wait', 'Overall_Rating', 'Recommend_Score', 'Nurse_Score', 'Doctor_Score']].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix, 
    annot = True, #shows coefficients in the box
    fmt='.2f', 
    cmap="coolwarm", 
    vmin=-1, 
    vmax=1, 
    linewidths=0.5, 
    square=True
)

plt.title("Correlation Heatmap", fontsize=16, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```
